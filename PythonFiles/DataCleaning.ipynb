```python
df["Kilometers Travelled"] = (
    df["Kilometers Travelled"]
    .str.replace("k km", "", regex=True)
    .astype(float) * 1000  
)
```

```python
df
```

```python
df["Ownership"] = df["Ownership"].str.extract("(\\d+)").astype(float)
df
```

```python
df["EMI"] = (
    df["EMI"]
    .str.replace("₹", "", regex=True)
    .str.replace("/m", "", regex=True)
    .str.replace(",", "", regex=True)
    .astype(float)
)
df
```

```python
df["Price"] = (
    df["Price"]
    .str.replace(" lakh", "", regex=True)
    .astype(float) * 100000  
)
df
```

```python
df = df.drop_duplicates()
df
```

```python
df.loc[:, "Company Name"] = df["Car Name"].str.split().str[0]
df
```
